```python




from transformers import EsmModel, EsmTokenizer
import torch
import numpy as np

# Charger le modèle (une seule fois)
print("Chargement du modèle ESM-2...")
model_name = "facebook/esm2_t6_8M_UR50D"
tokenizer = EsmTokenizer.from_pretrained(model_name)
model = EsmModel.from_pretrained(model_name)
model.eval()
print(" Modèle prêt !\n")

# Fonction simple pour obtenir l'embedding
def get_embedding(sequence):
    inputs = tokenizer(sequence, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
    # Moyenne sur les tokens (exclut CLS et SEP)
    embedding = outputs.last_hidden_state[0, 1:-1, :].mean(dim=0).numpy()
    return embedding

# Test avec une vraie globine (myoglobine de cheval)
seq_myoglobine = "MVLSEGEWQLVLHVWAKVEADVAGHGQDILIRLFKSHPETLEKFDRFKHLKTEAEMKASEDLKKHGVTVLTALGAILKKKGHHEAELKPLAQSHATKHKIPIKYLEFISDAIIHVLHSKHPGDFGADAQGAMNKALELFRKDMASNYKELGFQG"

# Test avec une protéine non-globine (exemple)
seq_autre = "MAEGEPPRVRRQPVPKKPPKPEVLPMGSDSSVEGLGRRTNRFQRERFDRLGKFSLVASGCGSSSRGSSSSDGSGSSGVDRVTSTSEARGASVSGISSCDQESAPAGGPALRAPGSWAPGPSPRRSKLRG"

print(" Génération des embeddings...")
emb_globine = get_embedding(seq_myoglobine)
emb_autre = get_embedding(seq_autre)

print(f"Embedding globine   : dimension {emb_globine.shape}")
print(f" Embedding autre     : dimension {emb_autre.shape}")

# Calculer la similarité
similarite = np.dot(emb_globine, emb_autre) / (np.linalg.norm(emb_globine) * np.linalg.norm(emb_autre))
print(f"\nSimilarité cosinus entre les deux: {similarite:.4f}")
print("   (1.0 = identiques, 0.0 = totalement différentes)")
```

```python
import re
from pathlib import Path
from Bio.PDB import PDBParser


dossier_data = TP_PATH / "data"
# parcours des proteines disponibles dans le jeu de données
# on recherche le fichier dans le répertoire
fichier_sequences = None
for chemin in [
    TP_PATH / "globin_all_nn.fst",
    TP_PATH / "data" / "globin_all_nn.fst",
    Path.cwd() / "globin_all_nn.fst",
    TP_PATH.parent / "globin_all_nn.fst"
]:
    if chemin.exists():
        fichier_sequences = chemin
        break

# fonctions à ne pas modifier, récupération des données

def extraire_depuis_fasta(fichier):
    """Extrait les codes et séquences depuis le fichier FASTA globin_all_nn.fst"""
    proteines = []
    code_actuel = ""
    sequence_actuelle = []
    
    with open(fichier, 'r', encoding='utf-8') as f:
        for ligne in f:
            ligne = ligne.strip()
            if not ligne:
                continue
            if ligne.startswith('>'):
                if code_actuel and sequence_actuelle:
                    proteines.append((code_actuel, ''.join(sequence_actuelle)))
                    sequence_actuelle = []
                # Extraction du code (nom de la proteine dans le fichier des sequences en AA)
                match = re.search(r'>[^|]+\|([A-Z0-9]+)', ligne)
                if match:
                    code_actuel = match.group(1)
                else:
                    match = re.search(r'>(\S+)', ligne)
                    code_actuel = match.group(1) if match else ligne[1:].split()[0]
            else:
                if code_actuel:
                    sequence_actuelle.append(ligne)
        if code_actuel and sequence_actuelle:
            proteines.append((code_actuel, ''.join(sequence_actuelle)))
    
    return proteines

def extraire_depuis_fichiers_pdb(dossier):
    """Extrait les codes depuis les noms des fichiers PDB"""
    codes = []
    for pdb_file in dossier.glob("AF-*.pdb"):
        match = re.search(r'AF-([A-Z0-9]+)-F1-model_v6\.pdb', pdb_file.name)
        if match:
            codes.append(match.group(1))
    return codes


# vérification du traitement de fichier

print("=========================")
print("CHARGEMENT DES PROTÉINES")
print("=========================")

# Chargement des séquences depuis le fichier FASTA "globin_all_nn.fst"
proteines_avec_sequences = []
if fichier_sequences:
    print(f"Fichier FASTA trouvé: {fichier_sequences.name}")
    proteines_avec_sequences = extraire_depuis_fasta(fichier_sequences)
    print(f"   → {len(proteines_avec_sequences)} protéines avec séquences")
else:
    print("Fichier FASTA non trouvé, utilisation des fichiers PDB uniquement")

# liste des fichiers PDB disponibles
print(f"\n Dossier data: {dossier_data}")
fichiers_pdb = list(dossier_data.glob("AF-*.pdb"))
print(f"   → {len(fichiers_pdb)} fichiers PDB trouvés")

# codes extraits depuis les fichiers PDB
codes_pdb = extraire_depuis_fichiers_pdb(dossier_data)
print(f"   → {len(codes_pdb)} codes extraits des fichiers PDB")


# création de classes pour gérer les données facilement

class Proteine:
    """Classe représentant une protéine avec accès facile"""
    """Pour ajouter un attribut : self._____ = variable """
    def __init__(self, code, sequence=None):
        self.code = code
        self.sequence = sequence
        self.nom_fichier = f"AF-{code}-F1-model_v6.pdb"
        self.chemin = dossier_data / self.nom_fichier
        self.fichier_existe = self.chemin.exists()
        self.longueur = len(sequence) if sequence else None
    
    def charger_structure(self):
        """Charge la structure PDB avec PDBParser"""
        if not self.fichier_existe:
            print(f"Fichier non trouvé: {self.chemin}")
            return None
        parser = PDBParser(PERMISSIVE=1)
        return parser.get_structure(id=self.code, file=self.chemin.open("r"))
    
    def __repr__(self):
        return f"Proteine('{self.code}', existe={self.fichier_existe})"

# Crée le dictionnaire principal d'accès
proteines = {}

# Ajoute les protéines avec séquences (priorité)
for code, seq in proteines_avec_sequences:
    proteines[code] = Proteine(code, seq)

# Ajoute aussi les protéines des fichiers PDB qui n'étaient pas dans le FASTA
for code in codes_pdb:
    if code not in proteines:
        proteines[code] = Proteine(code)
# On pourra coder ultérieurement une fonction qui ajoute la séquence en AA du fichier PDB dans le document FASTA plus tard.


# AFFICHAGE 

print("\n" + "======================")
print("RÉSUMÉ DES PROTÉINES CHARGÉES")
print("==================")

proteines_avec_fichier = sum(1 for p in proteines.values() if p.fichier_existe)
proteines_avec_seq = sum(1 for p in proteines.values() if p.sequence)

print(f"Total protéines: {len(proteines)}")
print(f"Avec fichier PDB: {proteines_avec_fichier}")
print(f"Avec séquence AA: {proteines_avec_seq}")

# Afficher l'ensemble des proteines
print("\n Premier aperçu (limite maximale de 200 modifiable):")
for i, (code, p) in enumerate(list(proteines.items())[:200], 1):
    statut = "trouvé" if p.fichier_existe else "fichier introuvable"
    seq_info = f"({p.longueur} aa)" if p.longueur else "sans séquence"
    print(f"   {i}. {code:<20} {statut} {p.nom_fichier:<45} {seq_info}")


# FONCTIONS D'ACCÈS (MODIFIABLES ULTERIEUREMENT)

def get_proteine(code):
    """Récupère une protéine par son code"""
    return proteines.get(code)

def get_proteines_avec_fichier():
    """Retourne la liste des protéines dont le fichier PDB existe"""
    return [p for p in proteines.values() if p.fichier_existe]

def get_proteines_avec_sequence():
    """Retourne la liste des protéines avec séquence AA DANS Globin_all_nn.fst"""
    return [p for p in proteines.values() if p.sequence]

def charger_proteine(code):
    """Charge directement la structure PDB d'une protéine"""
    p = get_proteine(code)
    if p:
        return p.charger_structure()
    print(f"ATTENTION Protéine {code} non trouvée")
    return None






# Afficher les premiers codes disponibles pour copier/coller par exemple
print("\n" + "======================================" )
print("CODES DISPONIBLES (pour copier/coller ou identifier la proteine)")
print("============================" )
codes_disponibles = [p.code for p in get_proteines_avec_fichier()]
if codes_disponibles:
    print(f"codes = {codes_disponibles[:20]}")
    if len(codes_disponibles) > 20:
        print(f"# ... et {len(codes_disponibles)-20} autres")
else:
    print("Aucun code disponible")

# Sauvegarde pour utilisation ultérieure
# il est possible d'ajouter des caracteristiques 

import json
with open(TP_PATH / "proteines_info.json", 'w') as f:
    json.dump({
        code: {
            'code': p.code,
            'nom_fichier': p.nom_fichier,
            'chemin': str(p.chemin),
            'fichier_existe': p.fichier_existe,
            'sequence' : p.sequence,
            'longueur': p.longueur
        }
        for code, p in proteines.items()
    }, f, indent=2)
print(f"\n Informations sauvegardées dans: {TP_PATH / 'proteines_info.json'}")

print("\n Utilisez les fonctions ci-dessous pour accéder aux protéines.")

```

```python

def compare_proteines(param1, param2):
    """
    Compare deux protéines et calcule les distances entre leurs diagrammes de persistance.
    
    Paramètres:
        param1: soit code (str) soit position (int) de la première protéine
        param2: soit code (str) soit position (int) de la deuxième protéine
    
    Retourne:
        dict: Dictionnaire contenant les caractéristiques et les distances etc.
    """
    
    # on récupère la protéine
    def get_proteine_from_param(param):
        """Convertit un paramètre (code ou position) en objet Proteine"""
        if isinstance(param, int):  # Position
            with open(TP_PATH / "positions_mapping.json", 'r') as f:
                mapping = json.load(f)
            code = mapping.get(str(param))
            if not code:
                print(f" Position {param} non trouvée")
                return None
            print(f" Position {param} → Code: {code}")
        else:  # Code
            code = param
        
        # Récupérer la Proteine
        p = get_proteine(code)
        if not p:
            print(f" Protéine {code} non trouvée")
            return None
        return p
    
    
    # fonction qui extrait les données d'une protéine
   
    def traiter_proteine(p):
        """Traite une protéine et retourne ses caractéristiques et son diagramme"""
        print(f"\n Traitement de {p.code}...")
        
        # Charger la structure
        parser = PDBParser(PERMISSIVE=1)
        protein = parser.get_structure(id=0, file=p.chemin.open("r"))
        
        # Extraire les positions des atomes CA
        whole_protein = list(protein)
        pos_ca = extract_ca_atom_pos(whole_protein)
        
        # Calcul du complexe de Delaunay-Cech
        cech_cs = gd.DelaunayCechComplex(points=pos_ca)
        cech_simplex_tree = cech_cs.create_simplex_tree()
        
        # Calcul du diagramme de persistance
        cech_simplex_tree.compute_persistence()
        
        # Extraire le diagramme de persistance en dimension 1
        diagram = cech_simplex_tree.persistence_intervals_in_dimension(1)
        
        # Convertir en tableau numpy pour wasserstein
        # Filtrer les points infinis et convertir en numpy array
        diagram_filtered = []
        for d in diagram:
            if d[1] != float('inf') and not np.isinf(d[0]) and not np.isinf(d[1]):
                diagram_filtered.append([d[0], d[1]])
        
        diagram_np = np.array(diagram_filtered) if diagram_filtered else np.array([]).reshape(0, 2)
        
        return {
            'code': p.code,
            'nom_fichier': p.nom_fichier,
            'sequence': p.sequence,
            'longueur': p.longueur,
            'nb_atomes_CA': len(pos_ca),
            'diagram': diagram_np,
            'diagram_raw': diagram,
            'simplex_tree': cech_simplex_tree
        }
    
    
    print("\n" + "=====================")
    print("COMPARAISON DE DEUX PROTÉINES")
    print("====================")
    
    p1 = get_proteine_from_param(param1)
    p2 = get_proteine_from_param(param2)
    
    if not p1 or not p2:
        print("Impossible de comparer : une ou deux protéines non trouvées")
        return None
    
    
    data1 = traiter_proteine(p1)
    data2 = traiter_proteine(p2)
    
    


    diagram1 = data1['diagram']
    diagram2 = data2['diagram']
    
    print("\n" + "===============================")
    print("DISTANCES ENTRE DIAGRAMMES DE PERSISTANCE")
    print("=========================")
    
    # Vérifier que les diagrammes ne sont pas vides
    if len(diagram1) > 0 and len(diagram2) > 0:
        try:
            # Distance de Wasserstein (ordre 2)
            wass_dist_2 = wasserstein_distance(diagram1, diagram2, order=2)
            print(f"Distance de Wasserstein (ordre 2) : {wass_dist_2:.4f}")
            
            # Distance de Wasserstein (ordre 1)
            wass_dist_1 = wasserstein_distance(diagram1, diagram2, order=1)
            print(f"Distance de Wasserstein (ordre 1) : {wass_dist_1:.4f}")
        except Exception as e:
            print(f"Erreur calcul Wasserstein: {e}")
            wass_dist_2 = None
            wass_dist_1 = None
    else:
        print("Un des diagrammes est vide, impossible de calculer Wasserstein")
        wass_dist_2 = None
        wass_dist_1 = None
    
    # Distance Bottleneck (toujours disponible)
    # même principe que la distance de wasserstein mais avec une distance L infinie
    # c'est d( x1 , x2 ) = max | x1 - x2 | ici.
    # le résultat est moins précis qu'une distance Lp
    try:
        # Pour bottleneck, on peut utiliser les diagrammes raw
        diagram1_raw = data1['diagram_raw']
        diagram2_raw = data2['diagram_raw']
        diagram1_raw = [d for d in diagram1_raw if d[1] != float('inf')]
        diagram2_raw = [d for d in diagram2_raw if d[1] != float('inf')]
        
        if len(diagram1_raw) > 0 and len(diagram2_raw) > 0:
            bottleneck_dist = gd.bottleneck_distance(diagram1_raw, diagram2_raw)
            print(f" Distance Bottleneck : {bottleneck_dist:.4f}")
        else:
            bottleneck_dist = None
            print(" Un des diagrammes est vide, impossible de calculer Bottleneck")
    except Exception as e:
        print(f" Erreur calcul Bottleneck: {e}")
        bottleneck_dist = None
    
    # Interprétation
    
        
    if wass_dist_2 is not None:
        print("\n INTERPRÉTATION (Wasserstein-2):")
        
        # Calcul du pourcentage de similitude (formule exponentielle décroissante)
        # Plus la distance est petite, plus la similitude est élevée
        if wass_dist_2 <= 0:
            pourcentage_similitude = 100.0
        else:
            # Utilisation d'une fonction exponentielle: similitude = 100 * e^(-distance / facteur)
            # 
            
            pourcentage_similitude = 100 * np.exp(-wass_dist_2 / 5)

        print(f"Pourcentage de similitude : {pourcentage_similitude:.1f}%")
    
        
        # Interprétation textuelle basée sur le pourcentage
        if pourcentage_similitude >= 80:
            print("Interprétation : Les deux protéines sont TRÈS SIMILAIRES")
           
        elif pourcentage_similitude >= 60:
            print("Interprétation : Les deux protéines sont SIMILAIRES")
            
        elif pourcentage_similitude >= 40:
            print(" Interprétation : Les deux protéines sont MODÉRÉMENT SIMILAIRES")
            
        elif pourcentage_similitude >= 20:
            print("Interprétation : Les deux protéines sont DIFFÉRENTES")
            
        else:
            print("Interprétation : Les deux protéines sont TRÈS DIFFÉRENTES")
            
        
        
        print("Plus la distance de wasserstein est petite, plus les protéines sont similaires")
    
    # ============================================
    # AFFICHAGE DES CARACTÉRISTIQUES COMPARÉES
    # ============================================
    print("\n" + "===============")
    print("CARACTÉRISTIQUES COMPARÉES")
    print("===============")
    
    # En-tête
    print(f"{'Caractéristique':<25} {'Protéine 1':<30} {'Protéine 2':<30}")
    print("---------------------------------")
    
    # Code
    print(f"{'Code':<25} {data1['code']:<30} {data2['code']:<30}")
    
    # Fichier
    print(f"{'Fichier PDB':<25} {data1['nom_fichier']:<30} {data2['nom_fichier']:<30}")
    
    # Longueur
    print(f"{'Longueur (aa)':<25} {data1['longueur']:<30} {data2['longueur']:<30}")
    
    # Nombre d'atomes CA
    print(f"{'Nb atomes CA':<25} {data1['nb_atomes_CA']:<30} {data2['nb_atomes_CA']:<30}")
    
    # Points dans le diagramme
    print(f"{'Points diagramme':<25} {len(data1['diagram_raw']):<30} {len(data2['diagram_raw']):<30}")
    
    # Séquence (aperçu)
    #seq1 = data1['sequence'][:50] + "..." if data1['sequence'] and len(data1['sequence']) > 50 else str(data1['sequence'])
    #seq2 = data2['sequence'][:50] + "..." if data2['sequence'] and len(data2['sequence']) > 50 else str(data2['sequence'])
    #print(f"{'Séquence (aperçu)':<25} {seq1:<30} {seq2:<30}")
    
    
    
    # ============================================
    # VISUALISATION DES DIAGRAMMES
    # ============================================
    print("\n" + "=====================")
    print("VISUALISATION DES DIAGRAMMES DE PERSISTANCE")
    print("=================")
    
    # Diagramme de la protéine 1
    print(f"\n Diagramme de persistance - {data1['code']}")
    if len(data1['diagram_raw']) > 0:
        fig, ax = plt.subplots(figsize=(10, 8))
        gd.plot_persistence_diagram(data1['diagram_raw'], axes=ax)
        ax.set_title(f'Diagramme de persistance - {data1["code"]}', fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.show()

    else:
        print("   (Diagramme vide)")
    
    # Diagramme de la protéine 2
    print(f"\n Diagramme de persistance - {data2['code']}")
    if len(data2['diagram_raw']) > 0:
        fig, ax = plt.subplots(figsize=(10, 8))
        gd.plot_persistence_diagram(data2['diagram_raw'], axes=ax)
        ax.set_title(f'Diagramme de persistance - {data2["code"]}', fontsize=14, fontweight='bold')
        plt.tight_layout()
        plt.show()
    else:
        print("   (Diagramme vide)")
    
    
        # Barcode de la protéine 1
    print(f"\n Barcode - {data1['code']}")
    if len(data1['diagram_raw']) > 0:
        fig, ax = plt.subplots(figsize=(12, 6))
        gd.plot_persistence_barcode(data1['diagram_raw'], axes=ax)
        ax.set_title(f'Barcode de persistance - {data1["code"]}', fontsize=14, fontweight='bold')
        ax.set_xlabel('Persistance')
        ax.set_ylabel('Points')
        plt.tight_layout()
        plt.show()
    else:
        print("   (Barcode vide)")
    
    # Barcode de la protéine 2
    print(f"\n Barcode - {data2['code']}")
    if len(data2['diagram_raw']) > 0:
        fig, ax = plt.subplots(figsize=(12, 6))
        gd.plot_persistence_barcode(data2['diagram_raw'], axes=ax)
        ax.set_title(f'Barcode de persistance - {data2["code"]}', fontsize=14, fontweight='bold')
        ax.set_xlabel('Persistance')
        ax.set_ylabel('Points')
        plt.tight_layout()
        plt.show()
    else:
        print("   (Barcode vide)")
    

    
    # Retourner les résultats - utile pour d'autres programmes de comparaison
    # qui pourraient utiliser ces données
    return {
        'proteine1': data1['code'],
        'proteine2': data2['code'],
        'longueur1': data1['longueur'],
        'longueur2': data2['longueur'],
        'points_diagram1': len(data1['diagram_raw']),
        'points_diagram2': len(data2['diagram_raw']),
        'wasserstein_2': wass_dist_2,
        'wasserstein_1': wass_dist_1,
        'bottleneck': bottleneck_dist,
        'diagram1': data1['diagram_raw'],
        'diagram2': data2['diagram_raw']
    }


# ============================================
# EXEMPLES D'UTILISATION
# ============================================

print("\n" + "=============")
print(" EXEMPLES D'UTILISATION")
print("=============")

print("""
# Comparer deux protéines par leur position
resultat = compare_proteines(1, 2)

# Comparer deux protéines par leur code
resultat = compare_proteines('A0A0A8WJC6', 'A0A0K3AT43')

# Récupérer uniquement la distance de Wasserstein
resultat = compare_proteines(1, 2)
if resultat and resultat['wasserstein_2']:
    print(f"Distance: {resultat['wasserstein_2']}")
""")


```

```python
def wasserstein_proteines(code1, code2):
    """
    Retourne la distance de Wasserstein entre deux proteines.
    Version simplifiee avec les codes.
    """
    from Bio.PDB import PDBParser
    import gudhi as gd
    from tp_pims.vizualize_filtration import extract_ca_atom_pos
    import numpy as np
    
    p1, p2 = get_proteine(code1), get_proteine(code2)
    
    parser = PDBParser(PERMISSIVE=1)
    prot1 = parser.get_structure(id=0, file=p1.chemin.open("r"))
    prot2 = parser.get_structure(id=0, file=p2.chemin.open("r"))
    
    ca1 = extract_ca_atom_pos(list(prot1))
    ca2 = extract_ca_atom_pos(list(prot2))
    
    cech1 = gd.DelaunayCechComplex(points=ca1)
    cech2 = gd.DelaunayCechComplex(points=ca2)
    
    st1 = cech1.create_simplex_tree()
    st2 = cech2.create_simplex_tree()
    
    st1.compute_persistence()
    st2.compute_persistence()
    
    # Extraire les diagrammes et convertir en numpy array
    diag1_raw = st1.persistence_intervals_in_dimension(1)
    diag2_raw = st2.persistence_intervals_in_dimension(1)
    
    # Filtrer et convertir en numpy array
    diag1 = np.array([[d[0], d[1]] for d in diag1_raw if d[1] != float('inf')])
    diag2 = np.array([[d[0], d[1]] for d in diag2_raw if d[1] != float('inf')])
    
    if len(diag1) == 0 or len(diag2) == 0:
        return None
    
    from gudhi.wasserstein import wasserstein_distance
    return wasserstein_distance(diag1, diag2, order=2)

# Test
dist = wasserstein_proteines('A0A0A8WJC6', 'A0A0K3AT43')
print(f"Distance de Wasserstein: {dist}")
```

```python
def find_min_non_zero(matrix):
    """
    Trouve la valeur minimale non nulle dans une matrice et ses coordonnees.
    
    Parametres:
        matrix: liste de listes contenant des nombres ou 'Err'
    
    Retourne:
        tuple: (min_value, i, j) ou (None, None, None) si aucun minimum non nul trouve
    """
    min_value = None
    min_i = None
    min_j = None
    
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            val = matrix[i][j]
            # Verifier si c'est un nombre, non nul, et pas 'Err'
            if isinstance(val, (int, float)) and val != 0 and val != 'Err':
                if min_value is None or val < min_value:
                    min_value = val
                    min_i = i
                    min_j = j
    
    return min_value, min_i, min_j
```

```python


def texte_vers_anytree(texte_arbre):
   
    lignes = texte_arbre.strip().split('\n')
    
    lignes = [l.rstrip(')').strip() for l in lignes if l.strip()]
    
    racine = None
    pile = []  # (node, niveau)
    
    for ligne in lignes:
        
        prefix = ligne[:ligne.find('├')] if '├' in ligne else \
                 ligne[:ligne.find('└')] if '└' in ligne else \
                 ligne[:ligne.find('│')] if '│' in ligne else ''
        
    
        niveau = prefix.count('│') + (prefix.count(' ') // 3)
        
        
        nom = ligne
        for sep in ['├── ', '└── ', '│   ']:
            if sep in ligne:
                nom = ligne.split(sep)[-1]
                break
        
        nom = nom.strip()
        
        # Créer le nœud
        node = Node(nom)
        
        if racine is None:
            racine = node
            pile = [(node, niveau)]
        else:
            # Remonter dans la pile jusqu'au parent
            while pile and pile[-1][1] >= niveau:
                pile.pop()
            
            if pile:
                parent = pile[-1][0]
                node.parent = parent
            else:
                node.parent = racine
            
            pile.append((node, niveau))
    
    return racine

#  texte 
texte_arbre = """
Metazoaire
├── Deuterostomiens
│   └── Chordés
│       ├── Céphalochordés
│       │   └── Amphioxus
│       │       ├── AF-Q17134-F1-model_v6.pdb 
│       │       ├── AF-O47431-F1-model_v6.pdb 
│       │       ├── AF-F2Q9Y0-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X8-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X7-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X6-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X5-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X4-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X3-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X2-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X1-F1-model_v6.pdb 
│       │       ├── AF-F2Q9X0-F1-model_v6.pdb 
│       │       ├── AF-F2Q9W9-F1-model_v6.pdb 
│       │       ├── AF-F2Q9W8-F1-model_v6.pdb 
│       │       ├── AF-F2Q9W7-F1-model_v6.pdb 
│       │       ├── AF-F2Q9W6-F1-model_v6.pdb 
│       │       ├── AF-C3YWU0-F1-model_v6.pdb 
│       │       └── AF-C3Y097-F1-model_v6.pdb 
│       ├── Urochordés
│       │   └── Cione_jaune
│       │       ├── AF-Q70T64-F1-model_v6.pdb 
│       │       ├── AF-Q70T63-F1-model_v6.pdb 
│       │       ├── AF-Q70T62-F1-model_v6.pdb 
│       │       ├── AF-Q70PH6-F1-model_v6.pdb 
│       │       ├── AF-Q9NL48-F1-model_v6.pdb 
│       │       ├── AF-Q4H3K6-F1-model_v6.pdb 
│       │       ├── AF-P61802-F1-model_v6.pdb 
│       │       └── AF-F6U472-F1-model_v6.pdb 
│       └── Vertébrés
│           ├── Agnathes (Petromyozontidae)
│           │   ├── Lamproie_marine
│           │   │   ├── AF-Q98980-F1-model_v6.pdb 
│           │   │   ├── AF-Q9PVD5-F1-model_v6.pdb 
│           │   │   ├── AF-Q9I9I3-F1-model_v6.pdb 
│           │   │   ├── AF-P09968-F1-model_v6.pdb 
│           │   │   ├── AF-P09967-F1-model_v6.pdb 
│           │   │   ├── AF-P02208-F1-model_v6.pdb 
│           │   │   ├── AF-A0AAJ7U215-F1-model_v6.pdb 
│           │   │   ├── AF-A0AAJ7TMY1-F1-model_v6.pdb 
│           │   │   ├── AF-A0AAJ7T5N9-F1-model_v6.pdb 
│           │   │   └── AF-A0AAJ7SV95-F1-model_v6.pdb 
│           │   └── Lamproie_de_planer
│           │       └── AF-A0A0A8WJC6-F1-model_v6.pdb 
│           └── Gnathostomes
│               ├── Chondrichtyens
│               │   ├── Squaliformes
│               │   │   └── Requin_du_groenland
│               │   │       ├── AF-Q8AXK2-F1-model_v6.pdb 
│               │   │       ├── AF-C0HJZ4-F1-model_v6.pdb 
│               │   │       ├── AF-C0HJZ3-F1-model_v6.pdb 
│               │   │       ├── AF-A0A2H4ZEZ2-F1-model_v6.pdb 
│               │   │       ├── AF-Q9NPG2-F1-model_v6.pdb 
│               │   │       ├── AF-Q8WY91-F1-model_v6.pdb 
│               │   │       ├── AF-Q8WWM9-F1-model_v6.pdb 
│               │   │       ├── AF-Q8N7X0-F1-model_v6.pdb 
│               │   │       ├── AF-Q6B0K9-F1-model_v6.pdb 
│               │   │       ├── AF-P69905-F1-model_v6.pdb 
│               │   │       ├── AF-P69892-F1-model_v6.pdb 
│               │   │       ├── AF-P69891-F1-model_v6.pdb 
│               │   │       ├── AF-P68871-F1-model_v6.pdb 
│               │   │       ├── AF-P09105-F1-model_v6.pdb 
│               │   │       ├── AF-P02144-F1-model_v6.pdb 
│               │   │       ├── AF-P02100-F1-model_v6.pdb 
│               │   │       ├── AF-P02042-F1-model_v6.pdb 
│               │   │       ├── AF-P02008-F1-model_v6.pdb 
│               │   │       ├── AF-A8MTQ0-F1-model_v6.pdb 
│               │   │       └── AF-A0AVK6-F1-model_v6.pdb 
│               │   └── Batoïdes
│               │       └── Torpille_marbrée
│               │           ├── AF-P20245-F1-model_v6.pdb 
│               │           └── AF-P20244-F1-model_v6.pdb 
│               └── Ostéichtyens
│                   ├── Actinoptérygiens
│                   │   └── Poisson_zèbre
│                   │       ├── AF-Q90487-F1-model_v6.pdb 
│                   │       ├── AF-Q90486-F1-model_v6.pdb 
│                   │       ├── AF-Q90485-F1-model_v6.pdb 
│                   │       ├── AF-Q90259-F1-model_v6.pdb 
│                   │       ├── AF-Q575S8-F1-model_v6.pdb 
│                   │       ├── AF-Q90YJ2-F1-model_v6.pdb 
│                   │       ├── AF-Q9DDJ0-F1-model_v6.pdb 
│                   │       ├── AF-Q8UUR3-F1-model_v6.pdb 
│                   │       ├── AF-Q7ZT21-F1-model_v6.pdb 
│                   │       ├── AF-Q6VN46-F1-model_v6.pdb 
│                   │       ├── AF-Q2XNF5-F1-model_v6.pdb 
│                   │       ├── AF-Q2WG80-F1-model_v6.pdb 
│                   │       ├── AF-P0DM64-F1-model_v6.pdb 
│                   │       ├── AF-O93548-F1-model_v6.pdb 
│                   │       ├── AF-O42145-F1-model_v6.pdb 
│                   │       └── AF-A3KNI8-F1-model_v6.pdb 
│                   └── Sarcoptérygiens
│                       └── Tétrapodes
│                           ├── Mammifères
│                           │   ├── Euthériens (placentaires)
│                           │   │   ├── Homo_sapiens
│                           │   │   └── Mus_musculus
│                           │   │       ├── AF-Q9ER97-F1-model_v6.pdb 
│                           │   │       ├── AF-Q9CX80-F1-model_v6.pdb 
│                           │   │       ├── AF-Q9CQE0-F1-model_v6.pdb 
│                           │   │       ├── AF-Q7TPG1-F1-model_v6.pdb 
│                           │   │       ├── AF-P06467-F1-model_v6.pdb 
│                           │   │       ├── AF-P04444-F1-model_v6.pdb 
│                           │   │       ├── AF-P04443-F1-model_v6.pdb 
│                           │   │       ├── AF-P04247-F1-model_v6.pdb 
│                           │   │       ├── AF-P02762-F1-model_v6.pdb 
│                           │   │       ├── AF-P02104-F1-model_v6.pdb 
│                           │   │       ├── AF-P02089-F1-model_v6.pdb 
│                           │   │       ├── AF-P02088-F1-model_v6.pdb 
│                           │   │       ├── AF-P01942-F1-model_v6.pdb 
│                           │   │       └── AF-G3UZ78-F1-model_v6.pdb 
│                           │   └── Métathériens (marsupiaux)
│                           │       └── Oppossum_gris
│                           │           ├── AF-Q95J88-F1-model_v6.pdb 
│                           │           ├── AF-Q9GL60-F1-model_v6.pdb 
│                           │           ├── AF-Q8HY34-F1-model_v6.pdb 
│                           │           ├── AF-Q3KN63-F1-model_v6.pdb 
│                           │           ├── AF-P49141-F1-model_v6.pdb 
│                           │           ├── AF-P48798-F1-model_v6.pdb 
│                           │           ├── AF-F7GH03-F1-model_v6.pdb 
│                           │           ├── AF-F6YXK6-F1-model_v6.pdb 
│                           │           ├── AF-F6YMJ4-F1-model_v6.pdb 
│                           │           ├── AF-F6WYU4-F1-model_v6.pdb 
│                           │           └── AF-F6WII6-F1-model_v6.pdb 
│                           └── Oiseaux
│                               ├── Gallus_gallus
│                               │   ├── AF-Q90864-F1-model_v6.pdb 
│                               │   ├── AF-Q5QSB1-F1-model_v6.pdb 
│                               │   ├── AF-P02197-F1-model_v6.pdb 
│                               │   ├── AF-P02128-F1-model_v6.pdb 
│                               │   ├── AF-P02127-F1-model_v6.pdb 
│                               │   ├── AF-P02112-F1-model_v6.pdb 
│                               │   ├── AF-P02007-F1-model_v6.pdb 
│                               │   ├── AF-P02001-F1-model_v6.pdb 
│                               │   └── AF-P01994-F1-model_v6.pdb 
│                               └── Aves (autre)
└── Protostomiens
    ├── Arthropodes
    │   ├── D._melanogaster (mouche)
    │   └── D._pseudoobscura
    │       ├── AF-Q3MQ62-F1-model_v6.pdb 
    │       ├── AF-Q24570-F1-model_v6.pdb 
    │       ├── AF-Q9VNE2-F1-model_v6.pdb 
    │       ├── AF-Q9I7P6-F1-model_v6.pdb 
    │       ├── AF-Q7KLI1-F1-model_v6.pdb 
    │       └── AF-Q4V6L1-F1-model_v6.pdb 
    └── Nématodes
        └── C._elegans
            ├── AF-Q23117-F1-model_v6.pdb 
            ├── AF-Q22663-F1-model_v6.pdb 
            ├── AF-Q21978-F1-model_v6.pdb 
            ├── AF-Q21618-F1-model_v6.pdb 
            ├── AF-Q20638-F1-model_v6.pdb 
            ├── AF-Q20623-F1-model_v6.pdb 
            ├── AF-Q20015-F1-model_v6.pdb 
            ├── AF-Q19601-F1-model_v6.pdb 
            ├── AF-Q18311-F1-model_v6.pdb 
            ├── AF-Q18086-F1-model_v6.pdb 
            ├── AF-Q17884-F1-model_v6.pdb 
            ├── AF-Q17727-F1-model_v6.pdb 
            ├── AF-Q09514-F1-model_v6.pdb 
            ├── AF-Q09240-F1-model_v6.pdb 
            ├── AF-Q966A3-F1-model_v6.pdb 
            ├── AF-Q9XXI2-F1-model_v6.pdb 
            ├── AF-Q9U2R5-F1-model_v6.pdb 
            ├── AF-Q9GYP4-F1-model_v6.pdb 
            ├── AF-Q8MPY8-F1-model_v6.pdb 
            ├── AF-Q7YZV1-F1-model_v6.pdb 
            ├── AF-Q7YWV3-F1-model_v6.pdb 
            ├── AF-P30627-F1-model_v6.pdb 
            ├── AF-O76629-F1-model_v6.pdb 
            ├── AF-O62089-F1-model_v6.pdb 
            ├── AF-O17824-F1-model_v6.pdb 
            ├── AF-H9G343-F1-model_v6.pdb 
            ├── AF-G5EGB6-F1-model_v6.pdb 
            ├── AF-G5EF00-F1-model_v6.pdb 
            ├── AF-E1NZ07-F1-model_v6.pdb 
            ├── AF-D0VWM4-F1-model_v6.pdb 
            ├── AF-C1P638-F1-model_v6.pdb 
            ├── AF-B6EU61-F1-model_v6.pdb 
            ├── AF-B3WFV2-F1-model_v6.pdb 
            ├── AF-B3GWE1-F1-model_v6.pdb 
            ├── AF-A8UWQ7-F1-model_v6.pdb 
            ├── AF-A3RMS5-F1-model_v6.pdb 
            └── AF-A0A0K3AT43-F1-model_v6.pdb 
"""

# Convertir
arbre = texte_vers_anytree(texte_arbre)

# Afficher l'arbre converti
print("\n" + "=" * 60)
print("ARBRE CONVERTI AVEC ANYTREE")
print("=" * 60)
for pre, fill, node in RenderTree(arbre):
    print(f"{pre}{node.name}")

# Statistiques
print("\n" + "=" * 60)
print("STATISTIQUES")
print("=" * 60)
print(f"Nombre total de nœuds : {sum(1 for _ in arbre.descendants) + 1}")
print(f"Nombre de feuilles : {sum(1 for _ in arbre.leaves)}")
print(f"Hauteur de l'arbre : {arbre.height}")

# Accès aux nœuds
print("\n" + "=" * 60)
print("ACCÈS AUX NŒUDS")
print("=" * 60)
print(f"Racine : {arbre.name}")
print(f"Enfants de la racine : {[c.name for c in arbre.children]}")
```

```python
# ============================================
# COMPARAISON DES PROTÉINES PAR EMBEDDINGS
# ============================================

import numpy as np
import pandas as pd
from scipy.spatial.distance import pdist, squareform, cosine
from scipy.stats import pearsonr, spearmanr, permutation_test
from sklearn.manifold import MDS
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

# 1. Générer les embeddings pour toutes les protéines
# ====================================================

def generer_embeddings_proteines(codes_proteines, batch_size=10):
    """
    Génère les embeddings ESM-2 pour une liste de protéines
    
    Paramètres:
    -----------
    codes_proteines : list
        Liste des codes Uniprot
    batch_size : int
        Taille des lots pour éviter les surcharges mémoire
    
    Retourne:
    ---------
    dict : Dictionnaire {code: embedding}
    """
    print(f" Génération des embeddings pour {len(codes_proteines)} protéines...")
    
    # Charger le modèle ESM-2 (déjà fait dans le notebook)
    # Si non chargé, décommenter les lignes suivantes:
    # from transformers import EsmModel, EsmTokenizer
    # model_name = "facebook/esm2_t6_8M_UR50D"
    # tokenizer = EsmTokenizer.from_pretrained(model_name)
    # model = EsmModel.from_pretrained(model_name)
    # model.eval()
    
    embeddings = {}
    codes_avec_sequence = []
    
    # Filtrer les protéines avec séquence
    for code in codes_proteines:
        p = get_proteine(code)
        if p and p.sequence:
            codes_avec_sequence.append(code)
    
    print(f"   → {len(codes_avec_sequence)} protéines avec séquence")
    
    # Générer les embeddings par lots
    for i in tqdm(range(0, len(codes_avec_sequence), batch_size), desc="Génération embeddings"):
        batch_codes = codes_avec_sequence[i:i+batch_size]
        
        for code in batch_codes:
            p = get_proteine(code)
            try:
                embedding = get_embedding(p.sequence)
                embeddings[code] = embedding
            except Exception as e:
                print(f"   Erreur pour {code}: {e}")
                continue
    
    print(f" Embeddings générés: {len(embeddings)} protéines")
    return embeddings

# 2. Calculer la matrice de distances à partir des embeddings
# ===========================================================

def calculer_matrice_distances_embeddings(embeddings, methode='cosine'):
    """
    Calcule la matrice de distances à partir des embeddings
    
    Paramètres:
    -----------
    embeddings : dict
        Dictionnaire {code: embedding}
    methode : str
        'cosine', 'euclidean', 'correlation'
    
    Retourne:
    ---------
    tuple: (matrice_distances, codes)
    """
    codes = list(embeddings.keys())
    n = len(codes)
    matrice = np.zeros((n, n))
    
    # Extraire les vecteurs
    vecteurs = np.array([embeddings[code] for code in codes])
    
    print(f"\n Calcul de la matrice de distances ({methode})...")
    
    if methode == 'cosine':
        # Normaliser les vecteurs
        normes = np.linalg.norm(vecteurs, axis=1, keepdims=True)
        vecteurs_norm = vecteurs / (normes + 1e-8)
        # Matrice de similarité cosinus
        similarite = vecteurs_norm @ vecteurs_norm.T
        # Convertir en distance (1 - similarité)
        matrice = 1 - similarite
        
    elif methode == 'euclidean':
        matrice = squareform(pdist(vecteurs, metric='euclidean'))
        
    elif methode == 'correlation':
        matrice = squareform(pdist(vecteurs, metric='correlation'))
    
    # Mettre la diagonale à 0
    np.fill_diagonal(matrice, 0)
    
    print(f" Matrice créée: {matrice.shape}")
    return matrice, codes

# 3. Test de Mantel avec permutations (Monte Carlo)
# =================================================

def mantel_test_embeddings(mat1, mat2, permutations=999, methode='pearson'):
    """
    Test de Mantel entre deux matrices de distances
    
    Paramètres:
    -----------
    mat1, mat2 : numpy arrays
        Matrices de distances
    permutations : int
        Nombre de permutations pour le test Monte Carlo
    methode : str
        'pearson' ou 'spearman'
    
    Retourne:
    ---------
    dict : Résultats du test
    """
    # Extraire le triangle supérieur
    n = len(mat1)
    triu_indices = np.triu_indices(n, k=1)
    vect1 = mat1[triu_indices]
    vect2 = mat2[triu_indices]
    
    # Corrélation observée
    if methode == 'pearson':
        obs_corr = np.corrcoef(vect1, vect2)[0, 1]
    else:
        from scipy.stats import spearmanr
        obs_corr, _ = spearmanr(vect1, vect2)
    
    # Test de permutation
    perm_corrs = np.zeros(permutations)
    
    for i in range(permutations):
        # Permuter les lignes/colonnes de la deuxième matrice
        perm_indices = np.random.permutation(n)
        mat2_perm = mat2[perm_indices][:, perm_indices]
        vect2_perm = mat2_perm[triu_indices]
        
        if methode == 'pearson':
            perm_corrs[i] = np.corrcoef(vect1, vect2_perm)[0, 1]
        else:
            perm_corrs[i], _ = spearmanr(vect1, vect2_perm)
    
    # P-value (test bilatéral)
    p_value = np.sum(np.abs(perm_corrs) >= np.abs(obs_corr)) / permutations
    
    # Z-score
    z_score = (obs_corr - np.mean(perm_corrs)) / (np.std(perm_corrs) + 1e-8)
    
    return {
        'coefficient': obs_corr,
        'p_value': p_value,
        'z_score': z_score,
        'permutations': permutations,
        'mean_perm': np.mean(perm_corrs),
        'std_perm': np.std(perm_corrs),
        'methode': methode
    }

# 4. Comparaison des différentes méthodes d'embedding
# ===================================================

def comparer_methodes_embedding(embeddings, codes, methode_reference='cosine'):
    """
    Compare différentes méthodes de calcul de distances à partir des embeddings
    """
    methodes = ['cosine', 'euclidean', 'correlation']
    resultats = {}
    
    print("\n" + "="*60)
    print("COMPARAISON DES MÉTHODES DE DISTANCE")
    print("="*60)
    
    # Calculer les matrices pour chaque méthode
    matrices = {}
    for methode in methodes:
        matrice, _ = calculer_matrice_distances_embeddings(embeddings, methode)
        matrices[methode] = matrice
    
    # Calculer les corrélations entre les méthodes
    methode_ref = methodes[0]
    for i, methode in enumerate(methodes[1:], 1):
        resultat = mantel_test_embeddings(matrices[methode_ref], matrices[methode], permutations=499)
        resultats[f"{methode_ref}_vs_{methode}"] = resultat
        print(f"\n{methode_ref} vs {methode}:")
        print(f"  Mantel r = {resultat['coefficient']:.4f} (p={resultat['p_value']:.4f})")
    
    return matrices, resultats

# 5. Clustering des protéines basé sur les embeddings
# ===================================================

def clustering_embeddings(matrice_distances, codes, n_clusters_range=range(2, 11)):
    """
    Effectue un clustering hiérarchique des protéines basé sur leurs embeddings
    """
    from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
    
    print("\n" + "="*60)
    print("CLUSTERING DES PROTÉINES")
    print("="*60)
    
    # Linkage pour clustering hiérarchique
    condensed_dist = squareform(matrice_distances, checks=False)
    linkage_matrix = linkage(condensed_dist, method='average')
    
    # Déterminer le nombre optimal de clusters par silhouette score
    print("\n Recherche du nombre optimal de clusters...")
    silhouette_scores = []
    
    for k in n_clusters_range:
        clusters = fcluster(linkage_matrix, k, criterion='maxclust')
        if len(set(clusters)) > 1:
            score = silhouette_score(matrice_distances, clusters, metric='precomputed')
            silhouette_scores.append(score)
            print(f"  k={k}: silhouette = {score:.4f}")
    
    if silhouette_scores:
        best_k = list(n_clusters_range)[np.argmax(silhouette_scores)]
        print(f"\n Meilleur nombre de clusters: {best_k} (silhouette = {max(silhouette_scores):.4f})")
    else:
        best_k = 5
        print(f"\n Nombre de clusters par défaut: {best_k}")
    
    # Assigner les clusters
    clusters = fcluster(linkage_matrix, best_k, criterion='maxclust')
    
    # Créer un DataFrame avec les résultats
    df_clusters = pd.DataFrame({
        'code': codes,
        'cluster': clusters
    })
    
    # Statistiques par cluster
    print("\n Statistiques par cluster:")
    for cluster_id in sorted(df_clusters['cluster'].unique()):
        n_proteines = len(df_clusters[df_clusters['cluster'] == cluster_id])
        print(f"  Cluster {cluster_id}: {n_proteines} protéines")
    
    return df_clusters, linkage_matrix

# 6. Visualisation MDS (Multidimensional Scaling)
# ===============================================

def visualisation_mds(matrice_distances, codes, df_clusters=None, figsize=(12, 10)):
    """
    Visualisation des protéines en 2D avec MDS
    """
    print("\n" + "="*60)
    print("VISUALISATION MDS")
    print("="*60)
    
    # MDS en 2 dimensions
    mds = MDS(n_components=2, dissimilarity='precomputed', random_state=42, normalized_stress='auto')
    coords = mds.fit_transform(matrice_distances)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    # Colorer par cluster si disponible
    if df_clusters is not None:
        # Mapper les clusters aux couleurs
        clusters = df_clusters.set_index('code')['cluster']
        unique_clusters = sorted(clusters.unique())
        colors = plt.cm.tab20(np.linspace(0, 1, len(unique_clusters)))
        color_map = {c: colors[i] for i, c in enumerate(unique_clusters)}
        
        for i, code in enumerate(codes):
            cluster = clusters.get(code, 0)
            color = color_map.get(cluster, 'gray')
            ax.scatter(coords[i, 0], coords[i, 1], c=[color], s=50, alpha=0.7)
        
        # Légende
        patches = [plt.Line2D([0], [0], marker='o', color='w', 
                              markerfacecolor=color_map[c], 
                              markersize=10, label=f'Cluster {c}') 
                   for c in unique_clusters]
        ax.legend(handles=patches, title='Clusters', loc='best')
    else:
        ax.scatter(coords[:, 0], coords[:, 1], s=50, alpha=0.7)
    
    ax.set_xlabel('MDS Dimension 1', fontsize=12)
    ax.set_ylabel('MDS Dimension 2', fontsize=12)
    ax.set_title('Visualisation des protéines par MDS (basée sur embeddings ESM-2)', fontsize=14)
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    return coords

# 7. Analyse de la robustesse des embeddings
# ==========================================

def analyse_robustesse(embeddings, n_iterations=100):
    """
    Analyse de la robustesse des embeddings par bootstrap
    """
    print("\n" + "="*60)
    print("ANALYSE DE ROBUSTESSE (BOOTSTRAP)")
    print("="*60)
    
    codes = list(embeddings.keys())
    n = len(codes)
    vecteurs = np.array([embeddings[code] for code in codes])
    
    # Bootstrap sur les dimensions des embeddings
    n_dim = vecteurs.shape[1]
    correlations = []
    
    for i in range(n_iterations):
        # Sélectionner aléatoirement 80% des dimensions
        n_sample = int(0.8 * n_dim)
        dim_indices = np.random.choice(n_dim, n_sample, replace=False)
        
        vecteurs_sample = vecteurs[:, dim_indices]
        
        # Normaliser
        normes = np.linalg.norm(vecteurs_sample, axis=1, keepdims=True)
        vecteurs_norm = vecteurs_sample / (normes + 1e-8)
        
        # Calculer la similarité
        sim = vecteurs_norm @ vecteurs_norm.T
        
        # Comparer avec la matrice complète
        vecteurs_full = vecteurs.copy()
        normes_full = np.linalg.norm(vecteurs_full, axis=1, keepdims=True)
        vecteurs_full_norm = vecteurs_full / (normes_full + 1e-8)
        sim_full = vecteurs_full_norm @ vecteurs_full_norm.T
        
        # Corrélation
        triu = np.triu_indices(n, k=1)
        corr = np.corrcoef(sim_full[triu], sim[triu])[0, 1]
        correlations.append(corr)
    
    print(f" Robustesse des embeddings (bootstrap {n_iterations} itérations):")
    print(f"   Corrélation moyenne: {np.mean(correlations):.4f}")
    print(f"   Écart-type: {np.std(correlations):.4f}")
    print(f"   Intervalle de confiance 95%: [{np.percentile(correlations, 2.5):.4f}, {np.percentile(correlations, 97.5):.4f}]")
    
    return correlations

# 8. Fonction principale de comparaison
# =====================================

def comparer_proteines_ia(codes_proteines=None, use_existing_embeddings=False, embeddings_dict=None):
    """
    Fonction principale pour comparer les protéines avec l'IA
    
    Paramètres:
    -----------
    codes_proteines : list
        Liste des codes à analyser (None = tous)
    use_existing_embeddings : bool
        Utiliser des embeddings existants
    embeddings_dict : dict
        Dictionnaire d'embeddings existants
    """
    print("\n" + "="*70)
    print(" COMPARAISON DES PROTÉINES PAR IA (EMBEDDINGS ESM-2)")
    print("="*70)
    
    # Obtenir les codes des protéines
    if codes_proteines is None:
        proteines_avec_fichier = get_proteines_avec_fichier()
        codes_proteines = [p.code for p in proteines_avec_fichier if p.sequence]
    
    print(f"\n Nombre de protéines à analyser: {len(codes_proteines)}")
    
    # Générer ou utiliser les embeddings
    if use_existing_embeddings and embeddings_dict:
        embeddings = embeddings_dict
    else:
        embeddings = generer_embeddings_proteines(codes_proteines)
    
    if len(embeddings) < 2:
        print(" Pas assez de protéines avec embeddings valides")
        return None
    
    # Calculer la matrice de distances
    matrice_distances, codes = calculer_matrice_distances_embeddings(embeddings, methode='cosine')
    
    # Sauvegarder la matrice
    df_matrice = pd.DataFrame(matrice_distances, index=codes, columns=codes)
    df_matrice.to_csv(TP_PATH / "matrice_distances_embeddings.csv")
    print(f"\n Matrice sauvegardée: {TP_PATH / 'matrice_distances_embeddings.csv'}")
    
    # Analyser la robustesse
    correlations_bootstrap = analyse_robustesse(embeddings, n_iterations=50)
    
    # Clustering
    df_clusters, linkage_matrix = clustering_embeddings(matrice_distances, codes)
    
    # Visualisation MDS
    coords = visualisation_mds(matrice_distances, codes, df_clusters)
    
    # Sauvegarder les clusters
    df_clusters.to_csv(TP_PATH / "clusters_proteines.csv", index=False)
    print(f"\n Clusters sauvegardés: {TP_PATH / 'clusters_proteines.csv'}")
    
    # Résumé des résultats
    print("\n" + "="*70)
    print(" RÉSUMÉ DES RÉSULTATS")
    print("="*70)
    print(f" {len(embeddings)} protéines analysées")
    print(f" Matrice de distances: {matrice_distances.shape}")
    print(f" Robustesse moyenne des embeddings: {np.mean(correlations_bootstrap):.3f}")
    print(f" Meilleur nombre de clusters: {df_clusters['cluster'].nunique()}")
    
    return {
        'embeddings': embeddings,
        'matrice_distances': matrice_distances,
        'codes': codes,
        'clusters': df_clusters,
        'coords_mds': coords,
        'robustesse': correlations_bootstrap
    }

# 9. Comparaison avec l'arbre de référence
# ========================================

def comparer_avec_arbre_reference(embeddings_resultats, arbre_reference):
    """
    Compare les résultats des embeddings avec l'arbre de référence
    """
    print("\n" + "="*60)
    print("COMPARAISON AVEC L'ARBRE DE RÉFÉRENCE")
    print("="*60)
    
    # Obtenir la matrice cophenétique de l'arbre
    mat_cophenique, labels_coph = obtenir_matrice_cophenetique(arbre_reference)
    
    # Créer DataFrames alignés
    df_emb = pd.DataFrame(
        embeddings_resultats['matrice_distances'],
        index=embeddings_resultats['codes'],
        columns=embeddings_resultats['codes']
    )
    df_coph = pd.DataFrame(mat_cophenique, index=labels_coph, columns=labels_coph)
    
    # Labels communs
    labels_communs = df_emb.index.intersection(df_coph.index)
    print(f"Labels communs: {len(labels_communs)}")
    
    if len(labels_communs) > 0:
        # Filtrer
        df_emb_filtre = df_emb.loc[labels_communs, labels_communs]
        df_coph_filtre = df_coph.loc[labels_communs, labels_communs]
        
        # Comparer avec Mantel
        resultats_comparaison = comparer_matrices_avance(
            df_emb_filtre.values,
            df_coph_filtre.values,
            permutations=499
        )
        afficher_comparaison(resultats_comparaison)
        
        return resultats_comparaison
    else:
        print(" Aucun label commun trouvé")
        return None

# 10. Exécution principale
# ========================

if __name__ == "__main__" or True:
    print("\n" + ""*35)
    print("ANALYSE DES PROTÉINES PAR IA - EMBEDDINGS ESM-2")
    print(""*35)
    
    # Exécuter l'analyse
    resultats_ia = comparer_proteines_ia()
    
    # Comparer avec l'arbre de référence (défini plus tôt)
    if 'arbre_reference' in dir() and resultats_ia:
        print("\n" + "="*60)
        print("COMPARAISON FINALE")
        print("="*60)
        comparer_avec_arbre_reference(resultats_ia, arbre_reference)
```

```python
# ============================================
# 1. GÉNÉRATION DES EMBEDDINGS POUR TOUTES LES PROTÉINES
# ============================================

import numpy as np
import pandas as pd
from transformers import EsmModel, EsmTokenizer
import torch
from tqdm import tqdm

def generer_tous_les_embeddings():
    """
    Génère les embeddings ESM-2 pour toutes les protéines avec séquence
    """
    print("\n" + "="*60)
    print("🚀 GÉNÉRATION DES EMBEDDINGS ESM-2")
    print("="*60)
    
    # 1. Charger le modèle ESM-2 (une seule fois)
    print("\n📥 Chargement du modèle ESM-2...")
    model_name = "facebook/esm2_t6_8M_UR50D"
    tokenizer = EsmTokenizer.from_pretrained(model_name)
    model = EsmModel.from_pretrained(model_name)
    model.eval()
    print("   ✅ Modèle chargé")
    
    # 2. Récupérer toutes les protéines avec séquence
    print("\n📋 Récupération des protéines...")
    proteines_avec_sequence = get_proteines_avec_sequence()
    print(f"   ✅ {len(proteines_avec_sequence)} protéines avec séquence")
    
    # 3. Générer les embeddings
    print("\n🧬 Génération des embeddings...")
    embeddings = {}
    
    for proteine in tqdm(proteines_avec_sequence, desc="Embeddings"):
        code = proteine.code
        sequence = proteine.sequence
        
        try:
            # Tokenizer la séquence
            inputs = tokenizer(sequence, return_tensors="pt", truncation=True, max_length=1022)
            
            # Générer l'embedding
            with torch.no_grad():
                outputs = model(**inputs)
            
            # Moyenne sur les tokens (excluant les tokens spéciaux)
            # last_hidden_state shape: (1, seq_len, hidden_size)
            embedding = outputs.last_hidden_state[0, 1:-1, :].mean(dim=0).numpy()
            
            embeddings[code] = embedding
            
        except Exception as e:
            print(f"   ⚠️ Erreur pour {code}: {e}")
            continue
    
    print(f"\n✅ Embeddings générés: {len(embeddings)} protéines")
    
    # 4. Afficher les dimensions
    if embeddings:
        premier_code = list(embeddings.keys())[0]
        print(f"   Dimension des embeddings: {embeddings[premier_code].shape}")
    
    return embeddings

# Exécuter la génération
embeddings = generer_tous_les_embeddings()
```

```python
# ============================================
# 1. DÉFINITION DE LA FONCTION MANQUANTE
# ============================================

def obtenir_matrice_cophenetique(racine):
    """
    Calcule et retourne la matrice cophenétique à partir d'un arbre phylogénétique.
    
    Paramètres:
    -----------
    racine : anytree.Node
        La racine de l'arbre
    
    Retourne:
    ---------
    mat_cophenique : numpy.ndarray
        Matrice cophenétique (distances entre toutes les paires de feuilles)
    labels : list
        Liste des labels des feuilles dans l'ordre de la matrice
    """
    from anytree import RenderTree
    
    # Récupérer toutes les feuilles
    feuilles = list(racine.leaves)
    n = len(feuilles)
    
    # Créer un dictionnaire pour associer chaque feuille à un indice
    indices = {feuille.name: i for i, feuille in enumerate(feuilles)}
    
    # Initialiser la matrice de distances
    distances = np.zeros((n, n))
    
    # Pour chaque paire de feuilles, calculer la distance cophenétique
    for i, f1 in enumerate(feuilles):
        for j, f2 in enumerate(feuilles):
            if i >= j:
                continue
            
            # Trouver l'ancêtre commun le plus proche
            ancetres_f1 = set()
            node = f1
            while node:
                ancetres_f1.add(node)
                node = node.parent
            
            node = f2
            ancetre_commun = None
            while node:
                if node in ancetres_f1:
                    ancetre_commun = node
                    break
                node = node.parent
            
            # La distance cophenétique est la profondeur de l'ancêtre commun
            if ancetre_commun:
                dist = ancetre_commun.depth
            else:
                dist = 0
            
            distances[i, j] = dist
            distances[j, i] = dist
    
    # Récupérer les labels dans l'ordre des feuilles
    labels = [feuille.name for feuille in feuilles]
    
    return distances, labels


# ============================================
# 2. COMPARAISON ARBRE vs EMBEDDINGS
# ============================================

import numpy as np
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt

print("\n" + "="*60)
print("🌳 CALCUL DE LA MATRICE COPHÉNIQUE DE L'ARBRE")
print("="*60)

# Calculer la matrice cophenétique
mat_cophenique, labels_coph = obtenir_matrice_cophenetique(arbre_reference)
print(f"✅ Matrice cophenétique: {mat_cophenique.shape}")
print(f"   Labels: {labels_coph[:5]}...")

# 2. Calculer la matrice de distance à partir des embeddings
# ==========================================================

print("\n🧬 CALCUL DE LA MATRICE DE DISTANCE DES EMBEDDINGS")
print("="*60)

# Extraire les codes des embeddings
codes_embeddings = list(embeddings.keys())
print(f"Codes embeddings: {len(codes_embeddings)}")
print(f"Premiers codes: {codes_embeddings[:5]}")

# Trouver les codes communs
codes_communs = [code for code in codes_embeddings if code in labels_coph]
print(f"\n🔍 Codes communs: {len(codes_communs)}/{len(codes_embeddings)}")

if len(codes_communs) == 0:
    print("\n⚠️ Attention: Aucun code commun trouvé!")
    print("   Les labels de l'arbre sont de la forme: 'AF-XXXX-F1-model_v6.pdb'")
    print("   Les codes des embeddings sont: sans le préfixe 'AF-' et suffixe")
    print("\n   Tentative de correction...")
    
    # Nettoyer les labels de l'arbre pour extraire seulement le code
    labels_coph_clean = []
    for label in labels_coph:
        # Extraire le code entre 'AF-' et '-F1-model_v6.pdb'
        if label.startswith('AF-') and '-F1-model_v6.pdb' in label:
            code_clean = label.replace('AF-', '').replace('-F1-model_v6.pdb', '')
            labels_coph_clean.append(code_clean)
        else:
            labels_coph_clean.append(label)
    
    # Re-trouver les codes communs
    codes_communs = [code for code in codes_embeddings if code in labels_coph_clean]
    print(f"   Après nettoyage: {len(codes_communs)} codes communs")

if len(codes_communs) > 0:
    # Créer un mapping des indices
    indices_coph = {label: i for i, label in enumerate(labels_coph)}
    
    # Extraire les indices correspondants dans la matrice cophenétique
    indices_coph_communs = []
    for code in codes_communs:
        # Trouver l'index dans labels_coph (avec ou sans préfixe)
        if code in labels_coph:
            idx = labels_coph.index(code)
        elif f"AF-{code}-F1-model_v6.pdb" in labels_coph:
            idx = labels_coph.index(f"AF-{code}-F1-model_v6.pdb")
        else:
            # Chercher par correspondance partielle
            for i, label in enumerate(labels_coph):
                if code in label:
                    idx = i
                    break
            else:
                continue
        indices_coph_communs.append(idx)
    
    # Extraire les vecteurs pour les codes communs
    vecteurs = np.array([embeddings[code] for code in codes_communs])
    
    # Normaliser les vecteurs pour la distance cosinus
    normes = np.linalg.norm(vecteurs, axis=1, keepdims=True)
    vecteurs_norm = vecteurs / (normes + 1e-8)
    
    # Matrice de similarité cosinus
    similarite = vecteurs_norm @ vecteurs_norm.T
    
    # Convertir en distance
    mat_embeddings = 1 - similarite
    np.fill_diagonal(mat_embeddings, 0)
    
    print(f"✅ Matrice embeddings: {mat_embeddings.shape}")
    
    # Extraire la sous-matrice cophenétique correspondante
    mat_cophenique_commune = mat_cophenique[np.ix_(indices_coph_communs, indices_coph_communs)]
    
    print(f"✅ Matrice cophenétique filtrée: {mat_cophenique_commune.shape}")
    
    # 4. Extraire les triangles supérieurs pour comparaison
    # =====================================================
    n = len(codes_communs)
    triu_indices = np.triu_indices(n, k=1)
    vect_emb = mat_embeddings[triu_indices]
    vect_coph = mat_cophenique_commune[triu_indices]
    
    print(f"\n📐 Nombre de paires comparées: {len(vect_emb)}")
    
    # 5. Calcul des corrélations
    # ==========================================================
    corr_pearson, p_pearson = pearsonr(vect_emb, vect_coph)
    corr_spearman, p_spearman = spearmanr(vect_emb, vect_coph)
    
    # 6. Test de Mantel simplifié
    # ==========================================================
    print("\n🔄 Calcul du test de Mantel...")
    
    obs_corr = np.corrcoef(vect_emb, vect_coph)[0, 1]
    permutations = 499
    perm_corrs = np.zeros(permutations)
    
    for i in range(permutations):
        perm_indices = np.random.permutation(n)
        mat_perm = mat_embeddings[perm_indices][:, perm_indices]
        vect_perm = mat_perm[triu_indices]
        perm_corrs[i] = np.corrcoef(vect_coph, vect_perm)[0, 1]
    
    p_mantel = np.sum(np.abs(perm_corrs) >= np.abs(obs_corr)) / permutations
    z_score = (obs_corr - np.mean(perm_corrs)) / (np.std(perm_corrs) + 1e-8)
    
    # 7. AFFICHAGE DES RÉSULTATS
    # ==========================================================
    print("\n" + "="*70)
    print("📊 RÉSULTATS DE LA COMPARAISON")
    print("="*70)
    
    print(f"\n📈 CORRÉLATIONS:")
    print(f"   • Pearson:  r = {corr_pearson:.4f}  (p = {p_pearson:.4f})")
    print(f"   • Spearman: ρ = {corr_spearman:.4f}  (p = {p_spearman:.4f})")
    
    print(f"\n🔄 TEST DE MANTEL (Monte Carlo):")
    print(f"   • Coefficient: r = {obs_corr:.4f}")
    print(f"   • P-value:     p = {p_mantel:.4f}")
    print(f"   • Z-score:     z = {z_score:.4f}")
    
    # 8. INTERPRÉTATION
    # ==========================================================
    print("\n" + "="*70)
    print("💡 INTERPRÉTATION")
    print("="*70)
    
    if p_mantel < 0.05:
        print("✅ Test de Mantel SIGNIFICATIF (p < 0.05)")
        if abs(obs_corr) > 0.7:
            print("   → Très forte similarité entre les deux matrices")
        elif abs(obs_corr) > 0.5:
            print("   → Forte similarité entre les deux matrices")
        elif abs(obs_corr) > 0.3:
            print("   → Similarité MODÉRÉE entre les deux matrices")
        else:
            print("   → Similarité FAIBLE mais significative")
    else:
        print("❌ Test de Mantel NON SIGNIFICATIF (p ≥ 0.05)")
    
    if obs_corr > 0:
        print(f"\n📈 Corrélation POSITIVE")
    else:
        print(f"\n📉 Corrélation NÉGATIVE")
    
    # 9. Visualisation
    # ==========================================================
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Distribution des permutations
    ax1.hist(perm_corrs, bins=30, alpha=0.7, color='steelblue', edgecolor='black')
    ax1.axvline(obs_corr, color='red', linestyle='--', linewidth=2, label=f'Observé: r={obs_corr:.4f}')
    ax1.axvline(np.mean(perm_corrs), color='green', linestyle='--', linewidth=2, label=f'Moyenne: {np.mean(perm_corrs):.4f}')
    ax1.set_xlabel('Corrélation de Mantel')
    ax1.set_ylabel('Fréquence')
    ax1.set_title(f'Distribution des permutations (n={permutations})')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Nuage de points
    ax2.scatter(vect_emb, vect_coph, alpha=0.5, s=10, c='steelblue')
    ax2.set_xlabel('Distance embeddings (cosinus)')
    ax2.set_ylabel('Distance cophenétique')
    ax2.set_title(f'Corrélation: r = {corr_pearson:.4f} (p = {p_pearson:.4f})')
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # 10. Résumé final
    print("\n" + "="*70)
    print("🎯 CONCLUSION FINALE")
    print("="*70)
    
    if p_mantel < 0.05:
        print(f"✅ Les embeddings ESM-2 capturent SIGNIFICATIVEMENT la structure phylogénétique")
        print(f"   (Mantel r = {obs_corr:.4f}, p = {p_mantel:.4f})")
    else:
        print(f"❌ Les embeddings ESM-2 ne capturent pas significativement la structure phylogénétique")
        print(f"   (Mantel r = {obs_corr:.4f}, p = {p_mantel:.4f})")
    
    print("="*70)
    
else:
    print("\n❌ Impossible de continuer: aucun code commun trouvé")
    print("   Vérifiez que les labels correspondent entre l'arbre et les embeddings")
```

```python
# ============================================
# 1. DÉFINITION DE LA FONCTION NEIGHBOR JOINING
# ============================================

import numpy as np
import pandas as pd

def neighbor_joining(dist_matrix):
    """
    Algorithme Neighbor Joining pour reconstruire un arbre phylogénétique
    
    Paramètres:
    -----------
    dist_matrix : array ou DataFrame
        Matrice de distances carrée, symétrique, diagonale nulle
    
    Retourne:
    ---------
    sorted_matrix : numpy.ndarray
        Matrice triée selon l'ordre déterminé par NJ
    order : list
        L'ordre des feuilles après l'algorithme
    """
    # Convertir en numpy array si c'est un DataFrame
    if isinstance(dist_matrix, pd.DataFrame):
        labels = dist_matrix.index.tolist()
        dist_matrix = dist_matrix.values
    else:
        labels = [f"Leaf_{i}" for i in range(len(dist_matrix))]
    
    n = len(dist_matrix)
    dist_matrix = np.array(dist_matrix, dtype=float)
    
    # Vérifications de base
    assert dist_matrix.shape[0] == dist_matrix.shape[1], "Matrice non carrée"
    assert np.allclose(dist_matrix, dist_matrix.T), "Matrice non symétrique"
    assert np.all(np.diag(dist_matrix) == 0), "Diagonale non nulle"
    
    # Copie pour ne pas modifier l'originale
    D = dist_matrix.copy()
    active_nodes = list(range(n))
    node_labels = labels.copy()
    
    # Dictionnaire pour suivre la hiérarchie
    clusters = {i: [i] for i in range(n)}
    
    while len(active_nodes) > 2:
        m = len(active_nodes)
        
        # Calculer les sommes des distances
        total_dist = np.array([sum(D[node, active_nodes]) for node in active_nodes])
        
        # Calculer la matrice Q
        Q = np.zeros((m, m))
        for i in range(m):
            for j in range(i+1, m):
                node_i, node_j = active_nodes[i], active_nodes[j]
                Q[i,j] = Q[j,i] = (m - 2) * D[node_i, node_j] - total_dist[i] - total_dist[j]
        
        # Trouver le minimum
        min_idx = np.unravel_index(np.argmin(Q), Q.shape)
        min_i, min_j = min_idx[0], min_idx[1]
        
        node_i, node_j = active_nodes[min_i], active_nodes[min_j]
        
        # Calculer les distances au nouveau nœud
        d_iu = (D[node_i, node_j] + (total_dist[min_i] - total_dist[min_j])/(m-2)) / 2
        d_ju = D[node_i, node_j] - d_iu
        
        # Créer nouveau nœud
        new_node = max(clusters.keys()) + 1
        
        # Fusionner les clusters
        clusters[new_node] = clusters[node_i] + clusters[node_j]
        
        # Mettre à jour la matrice D
        new_size = len(D) + 1
        new_D = np.zeros((new_size, new_size))
        new_D[:len(D), :len(D)] = D
        
        # Calculer les distances vers le nouveau nœud
        for node_k in active_nodes:
            if node_k not in [node_i, node_j]:
                new_D[new_node, node_k] = (D[node_i, node_k] + D[node_j, node_k] - D[node_i, node_j]) / 2
                new_D[node_k, new_node] = new_D[new_node, node_k]
        
        # Remplacer les nœuds
        active_nodes.remove(node_i)
        active_nodes.remove(node_j)
        active_nodes.append(new_node)
        D = new_D
    
    # Construire l'ordre final
    if len(active_nodes) == 2:
        node1, node2 = active_nodes
        order = clusters[node1] + clusters[node2]
    else:
        order = clusters[active_nodes[0]]
    
    # Vérifier que tous les éléments sont présents
    assert len(order) == n, f"Erreur: ordre a {len(order)} éléments, attendu {n}"
    
    # Trier la matrice
    sorted_matrix = dist_matrix[np.ix_(order, order)]
    
    # Restaurer les labels si nécessaire
    if isinstance(dist_matrix, pd.DataFrame) or labels:
        sorted_matrix = pd.DataFrame(
            sorted_matrix,
            index=[labels[i] for i in order],
            columns=[labels[i] for i in order]
        )
    
    return sorted_matrix, order


# ============================================
# 2. DÉFINITION DE LA FONCTION POUR LA MATRICE COPHÉNIQUE
# ============================================

def obtenir_matrice_cophenetique(racine):
    """
    Calcule et retourne la matrice cophenétique à partir d'un arbre phylogénétique.
    """
    from anytree import RenderTree
    
    # Récupérer toutes les feuilles
    feuilles = list(racine.leaves)
    n = len(feuilles)
    
    # Créer un dictionnaire pour associer chaque feuille à un indice
    indices = {feuille.name: i for i, feuille in enumerate(feuilles)}
    
    # Initialiser la matrice de distances
    distances = np.zeros((n, n))
    
    # Pour chaque paire de feuilles, calculer la distance cophenétique
    for i, f1 in enumerate(feuilles):
        for j, f2 in enumerate(feuilles):
            if i >= j:
                continue
            
            # Trouver l'ancêtre commun le plus proche
            ancetres_f1 = set()
            node = f1
            while node:
                ancetres_f1.add(node)
                node = node.parent
            
            node = f2
            ancetre_commun = None
            while node:
                if node in ancetres_f1:
                    ancetre_commun = node
                    break
                node = node.parent
            
            # La distance cophenétique est la profondeur de l'ancêtre commun
            if ancetre_commun:
                dist = ancetre_commun.depth
            else:
                dist = 0
            
            distances[i, j] = dist
            distances[j, i] = dist
    
    # Récupérer les labels dans l'ordre des feuilles
    labels = [feuille.name for feuille in feuilles]
    
    return distances, labels


# ============================================
# 3. COMPARAISON: ARBRE NJ (EMBEDDINGS) vs ARBRE RÉFÉRENCE
# ============================================

import numpy as np
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt

print("\n" + "="*70)
print("🔬 1. CALCUL DE LA MATRICE DE DISTANCE DES EMBEDDINGS")
print("="*70)

# Extraire les vecteurs pour tous les codes
codes_embeddings = list(embeddings.keys())
vecteurs = np.array([embeddings[code] for code in codes_embeddings])

# Normaliser pour la distance cosinus
normes = np.linalg.norm(vecteurs, axis=1, keepdims=True)
vecteurs_norm = vecteurs / (normes + 1e-8)

# Matrice de similarité cosinus
similarite = vecteurs_norm @ vecteurs_norm.T

# Convertir en distance
mat_embeddings_dist = 1 - similarite
np.fill_diagonal(mat_embeddings_dist, 0)

print(f"✅ Matrice de distance embeddings: {mat_embeddings_dist.shape}")
print(f"   Distance min: {mat_embeddings_dist.min():.4f}")
print(f"   Distance max: {mat_embeddings_dist.max():.4f}")
print(f"   Distance moyenne: {mat_embeddings_dist.mean():.4f}")


# 4. Appliquer Neighbor Joining sur la matrice des embeddings
# ============================================================

print("\n" + "="*70)
print("🌳 2. APPLICATION DE NEIGHBOR JOINING SUR LES EMBEDDINGS")
print("="*70)

# Créer un DataFrame avec les bons labels
df_embeddings = pd.DataFrame(mat_embeddings_dist, 
                              index=codes_embeddings, 
                              columns=codes_embeddings)

# Appliquer neighbor_joining
print("   Exécution de Neighbor Joining...")
matrice_nj_emb, order_emb = neighbor_joining(df_embeddings)

print(f"✅ Neighbor Joining terminé")
print(f"   Ordre des feuilles (premiers): {order_emb[:5]}...")
print(f"   Matrice NJ: {matrice_nj_emb.shape}")


# 5. Calculer la matrice cophenétique de l'arbre de référence
# ============================================================

print("\n" + "="*70)
print("🌳 3. CALCUL DE LA MATRICE COPHÉNIQUE DE L'ARBRE RÉFÉRENCE")
print("="*70)

# Utiliser l'arbre déjà chargé (variable 'arbre')
mat_cophenique_ref, labels_coph_ref = obtenir_matrice_cophenetique(arbre)

# Nettoyer les labels (enlever les préfixes/suffixes)
labels_coph_clean = []
for label in labels_coph_ref:
    if label.startswith('AF-') and '-F1-model_v6.pdb' in label:
        code_clean = label.replace('AF-', '').replace('-F1-model_v6.pdb', '')
        labels_coph_clean.append(code_clean)
    else:
        labels_coph_clean.append(label)

print(f"✅ Matrice cophenétique référence: {mat_cophenique_ref.shape}")
print(f"   Labels nettoyés (premiers): {labels_coph_clean[:5]}...")


# 6. Aligner les deux matrices
# ============================================================

print("\n" + "="*70)
print("🔗 4. ALIGNEMENT DES MATRICES")
print("="*70)

# Les labels dans l'ordre NJ
labels_nj = [codes_embeddings[i] for i in order_emb]

# Trouver les codes communs
codes_communs = [code for code in labels_nj if code in labels_coph_clean]
print(f"Codes communs: {len(codes_communs)}")

if len(codes_communs) == 0:
    print("❌ Aucun code commun trouvé!")
else:
    # Indices dans la matrice NJ
    indices_nj = [labels_nj.index(code) for code in codes_communs]
    
    # Indices dans la matrice cophenétique
    indices_coph = [labels_coph_clean.index(code) for code in codes_communs]
    
    # Extraire les sous-matrices
    mat_nj_filtre = matrice_nj_emb[np.ix_(indices_nj, indices_nj)]
    mat_coph_filtre = mat_cophenique_ref[np.ix_(indices_coph, indices_coph)]
    
    print(f"✅ Matrice NJ filtrée: {mat_nj_filtre.shape}")
    print(f"✅ Matrice cophenétique filtrée: {mat_coph_filtre.shape}")
    
    # Extraire les triangles supérieurs
    n = len(codes_communs)
    triu_indices = np.triu_indices(n, k=1)
    vect_nj = mat_nj_filtre[triu_indices]
    vect_coph = mat_coph_filtre[triu_indices]
    
    print(f"📐 Nombre de paires comparées: {len(vect_nj)}")
    
    # 7. Calcul des corrélations
    # ============================================================
    print("\n" + "="*70)
    print("📈 5. CALCUL DES CORRÉLATIONS")
    print("="*70)
    
    # Corrélation de Pearson
    corr_pearson, p_pearson = pearsonr(vect_nj, vect_coph)
    
    # Corrélation de Spearman
    corr_spearman, p_spearman = spearmanr(vect_nj, vect_coph)
    
    print(f"   • Pearson:  r = {corr_pearson:.4f}  (p = {p_pearson:.4f})")
    print(f"   • Spearman: ρ = {corr_spearman:.4f}  (p = {p_spearman:.4f})")
    
    # 8. Test de Mantel
    # ============================================================
    print("\n🔄 6. TEST DE MANTEL (Monte Carlo)")
    print("="*70)
    
    obs_corr = np.corrcoef(vect_nj, vect_coph)[0, 1]
    permutations = 499
    perm_corrs = np.zeros(permutations)
    
    for i in range(permutations):
        perm_indices = np.random.permutation(n)
        mat_perm = mat_nj_filtre[perm_indices][:, perm_indices]
        vect_perm = mat_perm[triu_indices]
        perm_corrs[i] = np.corrcoef(vect_coph, vect_perm)[0, 1]
    
    p_mantel = np.sum(np.abs(perm_corrs) >= np.abs(obs_corr)) / permutations
    z_score = (obs_corr - np.mean(perm_corrs)) / (np.std(perm_corrs) + 1e-8)
    
    print(f"   • Coefficient: r = {obs_corr:.4f}")
    print(f"   • P-value:     p = {p_mantel:.4f}")
    print(f"   • Z-score:     z = {z_score:.4f}")
    
    # 9. Interprétation
    # ============================================================
    print("\n" + "="*70)
    print("💡 INTERPRÉTATION")
    print("="*70)
    
    if p_mantel < 0.05:
        print("✅ Test de Mantel SIGNIFICATIF (p < 0.05)")
        if abs(obs_corr) > 0.7:
            print("   → Très forte similarité entre les deux arbres")
        elif abs(obs_corr) > 0.5:
            print("   → Forte similarité entre les deux arbres")
        elif abs(obs_corr) > 0.3:
            print("   → Similarité MODÉRÉE entre les deux arbres")
        else:
            print("   → Similarité FAIBLE mais significative")
    else:
        print("❌ Test de Mantel NON SIGNIFICATIF (p ≥ 0.05)")
    
    if obs_corr > 0:
        print(f"\n📈 Corrélation POSITIVE: Les deux arbres sont alignés")
    else:
        print(f"\n📉 Corrélation NÉGATIVE: Les arbres sont inversés")
    
    # 10. Visualisation
    # ============================================================
    print("\n" + "="*70)
    print("📊 7. VISUALISATION")
    print("="*70)
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    # Heatmap de la matrice NJ
    im1 = axes[0].imshow(mat_nj_filtre[:50, :50], cmap='viridis', aspect='auto')
    axes[0].set_title("Matrice NJ (embeddings)\n(50 premières protéines)")
    axes[0].set_xlabel("Protéines")
    axes[0].set_ylabel("Protéines")
    plt.colorbar(im1, ax=axes[0], label="Distance")
    
    # Heatmap de la matrice cophenétique
    im2 = axes[1].imshow(mat_coph_filtre[:50, :50], cmap='viridis', aspect='auto')
    axes[1].set_title("Matrice cophenétique\n(arbre référence)")
    axes[1].set_xlabel("Protéines")
    axes[1].set_ylabel("Protéines")
    plt.colorbar(im2, ax=axes[1], label="Distance")
    
    # Distribution des permutations
    axes[2].hist(perm_corrs, bins=30, alpha=0.7, color='steelblue', edgecolor='black')
    axes[2].axvline(obs_corr, color='red', linestyle='--', linewidth=2, 
                   label=f'Observé: r={obs_corr:.4f}')
    axes[2].axvline(np.mean(perm_corrs), color='green', linestyle='--', linewidth=2, 
                   label=f'Moyenne: {np.mean(perm_corrs):.4f}')
    axes[2].set_xlabel('Corrélation de Mantel')
    axes[2].set_ylabel('Fréquence')
    axes[2].set_title(f'Distribution des permutations (n={permutations})\np = {p_mantel:.4f}')
    axes[2].legend()
    axes[2].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Nuage de points
    plt.figure(figsize=(10, 8))
    plt.scatter(vect_nj, vect_coph, alpha=0.4, s=5, c='steelblue')
    plt.xlabel('Distance NJ (embeddings)', fontsize=12)
    plt.ylabel('Distance cophenétique (arbre référence)', fontsize=12)
    plt.title(f'Corrélation entre les deux arbres\nPearson r = {corr_pearson:.4f} (p={p_pearson:.4f})', 
              fontsize=14, fontweight='bold')
    plt.grid(True, alpha=0.3)
    
    # Ajouter une ligne de régression
    z = np.polyfit(vect_nj, vect_coph, 1)
    p = np.poly1d(z)
    plt.plot(np.sort(vect_nj), p(np.sort(vect_nj)), "r--", 
             label=f'Tendance: y = {z[0]:.4f}x + {z[1]:.4f}')
    plt.legend()
    
    plt.tight_layout()
    plt.show()
    
    # 11. Conclusion finale
    # ============================================================
    print("\n" + "="*70)
    print("🎯 CONCLUSION FINALE")
    print("="*70)
    
    print(f"""
    📊 RÉSUMÉ DE LA COMPARAISON:
    
    • Arbre NJ (embeddings) vs Arbre de référence
    • {len(codes_communs)} protéines communes
    • {len(vect_nj)} paires de distances comparées
    
    📈 RÉSULTATS:
    • Mantel test:     r = {obs_corr:.4f}  (p = {p_mantel:.4f})
    • Pearson:         r = {corr_pearson:.4f} (p = {p_pearson:.4f})
    • Spearman:        ρ = {corr_spearman:.4f} (p = {p_spearman:.4f})
    
    {'✅' if p_mantel < 0.05 else '❌'} CONCLUSION: Les deux arbres sont {'significativement' if p_mantel < 0.05 else 'non significativement'} similaires
    """)
    
    print("="*70)
```
